# Retail Sales Data Analysis

## Project Overview

This project focuses on exploring retail sales data using Python. The objective is to understand sales performance, identify popular product categories, analyze customer demographics, and discover monthly sales trends.

## Tools and Technologies

* Python
* Pandas
* Matplotlib
* Jupyter Notebook

## Dataset

The dataset contains 1,000 transactions and 9 columns, including transaction date, customer ID, gender, age, product category, quantity, price per unit, and total amount.

## Project Objectives

* Explore and understand the dataset.
* Check missing values and duplicate records.
* Analyze total sales and transaction values.
* Compare sales across product categories and genders.
* Explore monthly sales trends.
* Visualize the findings using graphs.



```python
import pandas as pd

df = pd.read_csv("retail_sales_dataset.csv")

df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
print("\nDataset Information:")
df.info()
```

```python
df["Date"] = pd.to_datetime(df["Date"])

print(df.dtypes)
```

```python
print("Total Sales:", df["Total Amount"].sum())

print("Average Transaction Amount:", df["Total Amount"].mean())

print("Highest Transaction Amount:", df["Total Amount"].max())

print("Lowest Transaction Amount:", df["Total Amount"].min())
```

```python
category_sales = df.groupby("Product Category")["Total Amount"].sum().sort_values(ascending=False)

print(category_sales)
```

```python
gender_sales = df.groupby("Gender")["Total Amount"].sum()

print(gender_sales)
```

```python
gender_sales = df.groupby("Gender")["Total Amount"].sum()

print(gender_sales)
```

```python
import matplotlib.pyplot as plt
```

```python
gender_sales.plot(kind="bar")

plt.title("Total Sales by Gender")
plt.xlabel("Gender")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
print("Average Customer Age:", df["Age"].mean())
print("Youngest Customer:", df["Age"].min())
print("Oldest Customer:", df["Age"].max())
```

```python
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total Amount"].sum()

print(monthly_sales)
```

```python
monthly_sales = df.groupby(df["Date"].dt.to_period("M"))["Total Amount"].sum()

print(monthly_sales)
```

```python
import matplotlib.pyplot as plt

monthly_sales.plot(kind="line", marker="o", figsize=(10, 5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
category_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
category_sales = df.groupby("Product Category")["Total Amount"].sum()
print(category_sales)
```

```python
category_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
gender_sales.plot(kind="pie", autopct="%1.1f%%", figsize=(6, 6))

plt.title("Sales Distribution by Gender")
plt.ylabel("Total Sales")
plt.show()
```

```python
df["Age"].plot(kind="hist", bins=10, figsize=(8, 5))

plt.title("Customer Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.tight_layout()
plt.show()
```

```python
quantity_sales = df.groupby("Product Category")["Quantity"].sum()

print(quantity_sales)
```

```python
quantity_sales.plot(kind="bar", figsize=(8, 5))

plt.title("Quantity Sold by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Total Quantity Sold")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
gender_category_sales = df.groupby(
    ["Gender", "Product Category"]
)["Total Amount"].sum()

print(gender_category_sales)
```

```python
gender_category_sales.unstack().plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Sales by Gender and Product Category")
plt.xlabel("Gender")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.legend(title="Product Category")
plt.tight_layout()
plt.show()
```

```python
quantity_amount = df.groupby("Product Category").agg({
    "Quantity": "sum",
    "Total Amount": "sum"
})

print(quantity_amount)
```

```python
quantity_amount.plot(
    kind="bar",
    figsize=(9, 5)
)

plt.title("Quantity Sold vs Total Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Value")
plt.xticks(rotation=0)
plt.legend(["Quantity Sold", "Total Sales"])
plt.tight_layout()
plt.show()
```

```python
highest_month = monthly_sales.idxmax()
lowest_month = monthly_sales.idxmin()

print("Highest Sales Month:", highest_month)
print("Highest Monthly Sales:", monthly_sales.max())

print("Lowest Sales Month:", lowest_month)
print("Lowest Monthly Sales:", monthly_sales.min())
```

```python
print("Total Transactions:", df["Transaction ID"].nunique())
print("Unique Customers:", df["Customer ID"].nunique())
print("Average Customer Age:", round(df["Age"].mean(), 2))
print("Total Quantity Sold:", df["Quantity"].sum())
```

```python
print("========== RETAIL SALES SUMMARY ==========")

print("Total Revenue:", df["Total Amount"].sum())

print("Best Sales Category:",
      df.groupby("Product Category")["Total Amount"].sum().idxmax())

print("Most Purchased Category:",
      df.groupby("Product Category")["Quantity"].sum().idxmax())

print("Average Transaction Value:",
      round(df["Total Amount"].mean(), 2))

print("Average Customer Age:",
      round(df["Age"].mean(), 2))

print("Total Items Sold:", df["Quantity"].sum())

print("==========================================")
```

## Conclusion

The exploratory data analysis of the retail sales dataset provided useful insights into sales performance, product categories, customer demographics, and monthly sales trends.

The dataset contains 1,000 transactions and 1,000 unique customer IDs, with total revenue of 456,000 and 2,514 items sold. The average transaction value is 456.

Electronics generated the highest revenue, totaling 156,905, while Clothing had the highest quantity sold, with 894 items. Female customers accounted for 232,840 in sales, compared with 223,160 for male customers. The average customer age was approximately 41 years.

The monthly sales analysis showed variations in sales throughout the period. These findings can help a business understand its sales performance and identify areas for further investigation.

Overall, this project demonstrates how Python, pandas, and Matplotlib can be used to explore retail data, visualize trends, and generate useful business insights.
